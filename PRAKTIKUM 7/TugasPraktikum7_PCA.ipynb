```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = data.target

print("Shape Data:", X.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
## A. Hitung Jumlah Komponen untuk 80%, 90%, dan 95%
from sklearn.decomposition import PCA

pca_80 = PCA(n_components=0.80)
pca_80.fit(X_scaled)

pca_90 = PCA(n_components=0.90)
pca_90.fit(X_scaled)

pca_95 = PCA(n_components=0.95)
pca_95.fit(X_scaled)

print("Komponen untuk 80% variance :", pca_80.n_components_)
print("Komponen untuk 90% variance :", pca_90.n_components_)
print("Komponen untuk 95% variance :", pca_95.n_components_)
```

```python
## B. Logistic Regression Tanpa PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

logreg = LogisticRegression(max_iter=5000)

logreg.fit(X_train, y_train)

y_pred = logreg.predict(X_test)

acc_no_pca = accuracy_score(y_test, y_pred)

print("Accuracy tanpa PCA:", acc_no_pca)
```

```python
## C. Logistic Regression dengan PCA 2 Komponen
pca_2 = PCA(n_components=2)

X_pca_2 = pca_2.fit_transform(X_scaled)

X_train, X_test, y_train, y_test = train_test_split(
    X_pca_2,
    y,
    test_size=0.2,
    random_state=42
)

logreg = LogisticRegression(max_iter=5000)

logreg.fit(X_train, y_train)

y_pred = logreg.predict(X_test)

acc_pca_2 = accuracy_score(y_test, y_pred)

print("Accuracy PCA 2 Komponen:", acc_pca_2)
```

```python
## D. Logistic Regression dengan PCA 10 Komponen
pca_10 = PCA(n_components=10)

X_pca_10 = pca_10.fit_transform(X_scaled)

X_train, X_test, y_train, y_test = train_test_split(
    X_pca_10,
    y,
    test_size=0.2,
    random_state=42
)

logreg = LogisticRegression(max_iter=5000)

logreg.fit(X_train, y_train)

y_pred = logreg.predict(X_test)

acc_pca_10 = accuracy_score(y_test, y_pred)

print("Accuracy PCA 10 Komponen:", acc_pca_10)
```

```python
##E. Tabel Perbandingan Akurasi
comparison = pd.DataFrame({
    "Model": [
        "Tanpa PCA",
        "PCA 2 Komponen",
        "PCA 10 Komponen"
    ],
    "Accuracy": [
        acc_no_pca,
        acc_pca_2,
        acc_pca_10
    ]
})

comparison
```
